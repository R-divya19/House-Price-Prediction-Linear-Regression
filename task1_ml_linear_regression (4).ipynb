# AI & ML Internship – Task 1
## House Price Prediction Using Linear Regression

**Dataset:** California Housing Dataset (scikit-learn)

**Objective:** Build and evaluate a Linear Regression model to predict house prices.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import pickle
print("All libraries imported successfully!")
```

## 2. Load the California Housing Dataset

```python
housing = fetch_california_housing(as_frame=True)
df = housing.frame
print("California Housing Dataset")
display(df.head())
print("Dataset loaded successfully!")
print("Total Rows and Columns:", df.shape)
```

## 3. Explore the Dataset

```python
print("Dataset Shape:")
print(df.shape)
print("\nDataset Columns:")
print(df.columns)
print("\nDataset Information:")
df.info()
```

```python
print("Statistical Summary")
display(df.describe())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

## 4. Exploratory Data Analysis (EDA)

```python
df.hist(figsize=(14, 10), bins=30, edgecolor='black')
plt.suptitle('California Housing - Feature Distributions', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
correlation = df.corr()
plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('California Housing Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
target_correlation = df.corr()["MedHouseVal"]
print("Correlation with House Value:")
print(target_correlation.sort_values(ascending=False))
```

## 5. Split the Dataset into Training and Testing Sets

```python
X = df.drop(columns=["MedHouseVal"])
y = df["MedHouseVal"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

## 6. Train the Linear Regression Model

```python
model = LinearRegression()
model.fit(X_train, y_train)
print("Linear Regression model trained successfully!")
```

## 7. Generate Predictions

```python
y_pred = model.predict(X_test)
print("Predictions generated successfully!")
print("\nFirst 10 predicted house values:")
print(y_pred[:10])
```

## 8. Evaluate the Model

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print(f"MAE: {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R2 Score: {r2:.4f}")
```

## 9. Actual vs. Predicted House Values

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.35)
line_min = min(y_test.min(), y_pred.min())
line_max = max(y_test.max(), y_pred.max())
plt.plot([line_min, line_max], [line_min, line_max], 'r--', label='Ideal Prediction')
plt.xlabel('Actual House Value (in $100,000)')
plt.ylabel('Predicted House Value (in $100,000)')
plt.title('Actual vs Predicted House Values')
plt.legend()
plt.tight_layout()
plt.show()
```

## 10. Residual Plot

```python
residuals = y_test - y_pred
plt.figure(figsize=(8, 5))
plt.scatter(y_pred, residuals, alpha=0.35)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel('Predicted House Value')
plt.ylabel('Residual (Actual - Predicted)')
plt.title('Residual Plot')
plt.tight_layout()
plt.show()
```

## 11. Save the Trained Model

```python
with open("california_house_price_model.pkl", "wb") as file:
    pickle.dump(model, file)
print("Model saved successfully!")
```

The Linear Regression model was trained and evaluated using the California Housing dataset. The model's performance was measured using MAE, RMSE and R². The visualizations compare actual and predicted house values and help inspect prediction errors.
